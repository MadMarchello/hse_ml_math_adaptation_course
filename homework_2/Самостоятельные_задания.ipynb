### Задание 1.3.1

Создать матрицу 8×8 и заполнить её в шахматном порядке нулями и единицами.

### Задание 2.3.1

Найти евклидово расстояние между двумя Series (точками) a и b, не используя встроенную формулу.

Точки в условии не заданы. Ниже a = (0, 0), b = (3, 4): расстояние равно 5.

```python
import numpy as np
import pandas as pd

a = pd.Series([0, 0])
b = pd.Series([3, 4])
distance = np.sqrt(((a - b) ** 2).sum())
print(distance)
```

## 3. Масштабирование

### Задание 3.3.2

Загрузить фрейм по ссылке `https://raw.githubusercontent.com/akmand/datasets/master/iris.csv`.

Нормализовать первый числовой признак `sepal_length_cm` минимаксным преобразованием, второй признак `sepal_width_cm` — z-масштабированием.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

iris = pd.read_csv('https://raw.githubusercontent.com/akmand/datasets/master/iris.csv')

iris['sepal_length_cm'] = MinMaxScaler().fit_transform(iris[['sepal_length_cm']])
iris['sepal_width_cm'] = StandardScaler().fit_transform(iris[['sepal_width_cm']])

display(iris.head())
print("sepal_length_cm: min =", iris["sepal_length_cm"].min(),
      "max =", iris["sepal_length_cm"].max())
print("sepal_width_cm: mean =", round(float(iris["sepal_width_cm"].mean()), 6),
      "std =", round(float(iris["sepal_width_cm"].std(ddof=0)), 6))
```
