```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Настройка графика
fig, ax = plt.subplots(figsize=(6, 11))
ax.set_aspect('equal')
ax.set_xlim(-0.5, 3.5)
ax.set_ylim(-0.5, 7.5)
ax.axis('off') # Отключаем оси

# ==========================================
# 1. ОТРИСОВКА СТЕН И КОМНАТ
# ==========================================

# Внешние стены (общий контур)
outer_wall = patches.Rectangle((0, 0), 3.0, 7.0, linewidth=4, edgecolor='#333333', facecolor='#F5F5F5')
ax.add_patch(outer_wall)

# Ванная комната (С/У)
bath = patches.Rectangle((0, 5.5), 1.4, 1.5, linewidth=2, edgecolor='#333333', facecolor='#D4E6F1')
ax.add_patch(bath)
ax.text(0.7, 6.25, "С/У\n3,5 м²", ha='center', va='center', fontsize=10, weight='bold')

# Прихожая
hall = patches.Rectangle((1.4, 5.5), 1.6, 1.5, linewidth=2, edgecolor='#333333', facecolor='#EAECEE')
ax.add_patch(hall)
ax.text(2.2, 6.25, "Прихожая\n3,5 м²", ha='center', va='center', fontsize=10, weight='bold')

# Кухня-ниша
kitchen = patches.Rectangle((0, 3.2), 1.4, 2.3, linewidth=2, edgecolor='#333333', facecolor='#FDEBD0')
ax.add_patch(kitchen)
ax.text(0.7, 4.35, "Кухня-ниша\n5,0 м²", ha='center', va='center', fontsize=10, weight='bold')

# Гостиная (центральная часть)
living = patches.Rectangle((1.4, 3.2), 1.6, 2.3, linewidth=2, edgecolor='#333333', facecolor='#D5F5E3')
ax.add_patch(living)
ax.text(2.2, 4.35, "Гостиная\n9,0 м²", ha='center', va='center', fontsize=10, weight='bold')

# ==========================================
# 2. ЗОНИРОВАНИЕ: СПАЛЬНЯ-НИША И ШКАФ
# ==========================================

# Зона спальни (ниша)
bedroom = patches.Rectangle((0, 0), 3.0, 3.2, linewidth=2, edgecolor='#333333', facecolor='#E8DAEF', alpha=0.5, linestyle='--')
ax.add_patch(bedroom)
ax.text(1.5, 3.0, "СПАЛЬНЯ-НИША\n(Зона отдыха)", ha='center', va='center', fontsize=11, weight='bold', color='#6C3483')

# Шкаф-перегородка (отделяет спальню от кухни/гостиной)
# Оставляем проход слева (0.6 м) для доступа к кухне и окну
wardrobe = patches.Rectangle((0.6, 2.6), 2.4, 0.6, linewidth=2, edgecolor='#333333', facecolor='#A0522D')
ax.add_patch(wardrobe)
ax.text(1.8, 2.9, "ШКАФ (2.4 x 0.6 м)", ha='center', va='center', fontsize=9, color='white', weight='bold')

# Проход в спальню
ax.annotate('Проход\n0.6 м', xy=(0.3, 2.9), xytext=(0.3, 3.5), 
            arrowprops=dict(facecolor='black', arrowstyle='->'), ha='center', fontsize=9)

# ==========================================
# 3. РАССТАНОВКА МЕБЕЛИ
# ==========================================

# --- Спальня ---
# Кровать (Двуспальная, 1.6 x 2.0 м)
bed = patches.Rectangle((0.2, 0.2), 1.6, 2.0, linewidth=1, edgecolor='#555555', facecolor='#FFFFFF')
ax.add_patch(bed)
# Подушки
pillow1 = patches.Rectangle((0.3, 1.9), 0.6, 0.3, linewidth=1, edgecolor='#999999', facecolor='#F2F3F4')
pillow2 = patches.Rectangle((1.1, 1.9), 0.6, 0.3, linewidth=1, edgecolor='#999999', facecolor='#F2F3F4')
ax.add_patch(pillow1)
ax.add_patch(pillow2)
ax.text(1.0, 1.1, "Кровать\n160х200", ha='center', va='center', fontsize=8)

# Прикроватная тумба
nightstand = patches.Rectangle((2.0, 1.9), 0.4, 0.4, linewidth=1, edgecolor='#555555', facecolor='#FFFFFF')
ax.add_patch(nightstand)

# --- Кухня ---
# Кухонный гарнитур
kitchen_set = patches.Rectangle((0, 3.2), 0.6, 2.3, linewidth=1, edgecolor='#555555', facecolor='#D3D3D3')
ax.add_patch(kitchen_set)
ax.text(0.3, 4.35, "Кухня", ha='center', va='center', rotation=90, fontsize=9)

# --- Гостиная ---
# Диван
sofa = patches.Rectangle((2.3, 3.4), 0.7, 1.6, linewidth=1, edgecolor='#555555', facecolor='#BDC3C7')
ax.add_patch(sofa)
ax.text(2.65, 4.2, "Диван", ha='center', va='center', rotation=90, fontsize=9)

# Стол (журнальный/обеденный)
table = patches.Rectangle((1.5, 3.4), 0.6, 0.6, linewidth=1, edgecolor='#555555', facecolor='#FFFFFF')
ax.add_patch(table)
ax.text(1.8, 3.7, "Стол", ha='center', va='center', fontsize=8)

# --- Ванная комната ---
# Ванна
bathtub = patches.Rectangle((0.1, 5.6), 0.7, 1.3, linewidth=1, edgecolor='#555555', facecolor='#FFFFFF')
ax.add_patch(bathtub)
# Раковина
sink = patches.Rectangle((1.0, 6.5), 0.3, 0.4, linewidth=1, edgecolor='#555555', facecolor='#FFFFFF')
ax.add_patch(sink)
# Унитаз
toilet = patches.Ellipse((1.1, 5.9), width=0.3, height=0.4, linewidth=1, edgecolor='#555555', facecolor='#FFFFFF')
ax.add_patch(toilet)

# ==========================================
# 4. ОКНА И ДВЕРИ
# ==========================================

# Окно (внизу)
window = patches.Rectangle((0.5, -0.1), 2.0, 0.1, linewidth=2, edgecolor='blue', facecolor='#AED6F1')
ax.add_patch(window)
ax.text(1.5, -0.3, "Окно", ha='center', va='center', fontsize=10, color='blue')

# Входная дверь (вверху справа)
door = patches.Arc((2.8, 7.0), 1.0, 1.0, angle=180, theta1=0, theta2=90, linewidth=2, color='#333333')
ax.add_patch(door)

# Дверь в ванную
bath_door = patches.Arc((1.4, 5.5), 0.7, 0.7, angle=0, theta1=180, theta2=270, linewidth=2, color='#333333')
ax.add_patch(bath_door)

# Заголовок
plt.title("Планировка студии 21 м² (Вариант 1: Спальня-ниша у окна)", fontsize=14, pad=20, weight='bold')

plt.tight_layout()
plt.show()
```
